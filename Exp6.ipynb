### Experiment 06 Implement Supervised and Unsupervised Neural Network Learning Techniques


```python

import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Dataset: student study hours and attendance
X = np.array([
    [1, 40],
    [2, 45],
    [3, 50],
    [4, 60],
    [5, 70],
    [6, 80],
    [7, 90],
    [8, 95]
])

# -------------------------------
# 1. Supervised Learning
# -------------------------------

# Labeled outputs: 0 = Fail, 1 = Pass
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = MLPClassifier(
    hidden_layer_sizes=(5,),
    activation="relu",
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)

model.fit(X_scaled, y)

predictions = model.predict(X_scaled)

print("SUPERVISED LEARNING")
print("Actual outputs:   ", y)
print("Predicted outputs:", predictions)

# Predict for a new student
new_student = np.array([[6, 75]])
new_student_scaled = scaler.transform(new_student)

result = model.predict(new_student_scaled)
print("New student prediction:",
      "Pass" if result[0] == 1 else "Fail")


# -------------------------------
# 2. Unsupervised Learning
# -------------------------------

# No labels are provided to the clustering algorithm
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

print("\nUNSUPERVISED LEARNING")
print("Cluster assignments:", clusters)
print("Cluster centers:")
print(kmeans.cluster_centers_)

print("\nLearning techniques implemented successfully.")
```

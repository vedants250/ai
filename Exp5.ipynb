### Experiment 05 Implement Perceptron Learning for Logic Gates (AND, OR, and XOR)


```python

import numpy as np

# Perceptron Learning Algorithm
class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=20):
        self.lr = learning_rate
        self.epochs = epochs

    def train(self, X, y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        for _ in range(self.epochs):
            for inputs, target in zip(X, y):
                net = np.dot(inputs, self.weights) + self.bias
                output = 1 if net > 0 else 0

                error = target - output
                self.weights += self.lr * error * inputs
                self.bias += self.lr * error

    def predict(self, X):
        net = np.dot(X, self.weights) + self.bias
        return (net > 0).astype(int)


# Input combinations
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Logic gate outputs
gates = {
    "AND": np.array([0, 0, 0, 1]),
    "OR":  np.array([0, 1, 1, 1]),
    "XOR": np.array([0, 1, 1, 0])
}

# Train and test each gate
for name, y in gates.items():
    print(f"\n{name} Gate:")

    if name == "XOR":
        print("A single perceptron cannot learn XOR.")
        print("Expected:", y)
        print("Reason: XOR is not linearly separable.")
        continue

    model = Perceptron()
    model.train(X, y)
    predictions = model.predict(X)

    print("Input  Expected  Predicted")
    for inputs, expected, predicted in zip(X, y, predictions):
        print(f"{inputs}      {expected}         {predicted}")

    print("Weights:", model.weights)
    print("Bias:", model.bias)
```

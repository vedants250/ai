### Experiment 04 Implement Adversarial Search Using the Min-max Algorithm


```python

# Experiment 04: Min-Max Algorithm

def minimax(depth, node_index, is_max, scores, max_depth):
    # If leaf node is reached
    if depth == max_depth:
        return scores[node_index]

    if is_max:
        return max(
            minimax(depth + 1, node_index * 2, False,
                    scores, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False,
                    scores, max_depth)
        )
    else:
        return min(
            minimax(depth + 1, node_index * 2, True,
                    scores, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True,
                    scores, max_depth)
        )


# Leaf node scores
scores = [3, 5, 2, 9, 12, 5, 23, 23]

# Calculate tree depth
import math
max_depth = int(math.log2(len(scores)))

# Find the optimal value
optimal_value = minimax(0, 0, True, scores, max_depth)

print("Leaf Node Scores:", scores)
print("Optimal Value using Min-Max:", optimal_value)
```

### Experiment 09 Implement a Fuzzy Logic Controller for Real-World Applications


```python

# Experiment: Fuzzy Logic Fan Controller

# Install required library
!pip install -q scikit-fuzzy

import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

# Define input and output variables
temperature = ctrl.Antecedent(
    np.arange(0, 41, 1), 'temperature'
)

fan_speed = ctrl.Consequent(
    np.arange(0, 101, 1), 'fan_speed'
)

# Define temperature membership functions
temperature['low'] = fuzz.trimf(
    temperature.universe, [0, 0, 20]
)

temperature['medium'] = fuzz.trimf(
    temperature.universe, [10, 22, 34]
)

temperature['high'] = fuzz.trimf(
    temperature.universe, [25, 40, 40]
)

# Define fan speed membership functions
fan_speed['slow'] = fuzz.trimf(
    fan_speed.universe, [0, 0, 50]
)

fan_speed['moderate'] = fuzz.trimf(
    fan_speed.universe, [25, 50, 75]
)

fan_speed['fast'] = fuzz.trimf(
    fan_speed.universe, [50, 100, 100]
)

# Define fuzzy rules
rule1 = ctrl.Rule(
    temperature['low'], fan_speed['slow']
)

rule2 = ctrl.Rule(
    temperature['medium'], fan_speed['moderate']
)

rule3 = ctrl.Rule(
    temperature['high'], fan_speed['fast']
)

# Create fuzzy control system
fan_control = ctrl.ControlSystem(
    [rule1, rule2, rule3]
)

controller = ctrl.ControlSystemSimulation(
    fan_control
)

# Get input from the user
temp = float(input("Enter room temperature (0-40 C): "))

# Validate input and calculate output
if not 0 <= temp <= 40:
    print("Temperature must be between 0 and 40 C.")

else:
    controller.input['temperature'] = temp

    # Compute fuzzy output
    controller.compute()

    speed = controller.output['fan_speed']

    # Display result
    print("\nFuzzy Logic Fan Controller")
    print(f"Room Temperature: {temp:.1f} C")
    print(f"Recommended Fan Speed: {speed:.2f}%")
```

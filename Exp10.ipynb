### Experiment 10 Implement a Hybrid Neuro-Fuzzy System


```python

import numpy as np

# Training data: temperature and corresponding fan speed
X = np.array([10, 15, 20, 25, 30, 35, 40], dtype=float)
Y = np.array([10, 20, 30, 50, 70, 85, 100], dtype=float)

# Define Gaussian membership function
def gaussian(x, center, sigma):
    return np.exp(-((x - center) ** 2) / (2 * sigma ** 2))

# Fuzzy rule centers
centers = np.array([10, 20, 30, 40], dtype=float)
sigma = 10.0

# Generate normalized fuzzy rule strengths
def get_features(x):
    memberships = np.array([
        gaussian(x, c, sigma) for c in centers
    ])

    normalized = memberships / np.sum(memberships)

    # First-order Sugeno model:
    # each rule output is a*x + b
    return np.concatenate([
        normalized * x,
        normalized
    ])

# Construct feature matrix
A = np.array([get_features(x) for x in X])

# Train consequent parameters using least squares
parameters, _, _, _ = np.linalg.lstsq(A, Y, rcond=None)

# Predict output for a new temperature
test_temperature = float(
    input("Enter temperature (10-40 C): ")
)

if not 10 <= test_temperature <= 40:
    print("Please enter a temperature between 10 and 40 C.")
else:
    prediction = get_features(test_temperature) @ parameters

    print("\nHybrid Neuro-Fuzzy System")
    print("Temperature:", test_temperature, "C")
    print("Predicted Fan Speed:", round(
        float(np.clip(prediction, 0, 100)), 2
    ), "%")

    print("\nTraining Results:")
    for x, actual in zip(X, Y):
        predicted = get_features(x) @ parameters
        print(
            f"Temperature: {x:.0f} C | "
            f"Actual: {actual:.1f}% | "
            f"Predicted: {predicted:.2f}%"
        )
```
